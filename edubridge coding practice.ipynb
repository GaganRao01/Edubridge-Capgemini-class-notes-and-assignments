```python
def linearsearch(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return arr[i]
    return -1

arr = [1, 2, 3, 5455, 3, 22]
target = 22
print(linearsearch(arr, target))
```

```python
def binarysearch(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return arr[mid]

        elif arr[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return -1

arr = [1,2,3,4,5,6,7,8,9]
target = 8
print(binarysearch(arr, target))
```

```python
def bubblesort(arr):
    for i in range(len(arr)-1):
        for j in range(len(arr)-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr
arr = [1,22,3,444,5,77,4,21]
print(bubblesort(arr))
```

```python
# The total number of rows we want in our pyramid.
size = 4

# This is the main loop that controls the rows.
# It will execute its block of code 4 times.
for row in range(1, size + 1):

    # --- PART 1: PRINT THE LEADING SPACES ---
    # This loop prints the necessary spaces for the current row.
    num_spaces_to_print = size - row
    for space_loop_count in range(num_spaces_to_print):
        print(" ", end="")

    # --- PART 2: PRINT THE STARS ---
    # This loop prints the stars for the current row.
    num_stars_to_print = row
    for star_loop_count in range(num_stars_to_print):
        print("* ", end="")

    # --- PART 3: MOVE TO THE NEXT LINE ---
    # The row is finished, so we print a newline to start the next row.
    print()
```

```python
def ope(tokens:list[str])->int:
    stack =[]
    for token in tokens:
        if token in "+-*/":
            num2= stack.pop()
            num1 = stack.pop()

            if token == "+":
                result = num1+num2
            elif token == "-":
                result = num1-num2
            elif token == "*":
                result = num1*num2
            else :
                result = int(num1/2)
            stack.append(result)
        else:
            stack.append(int(token))
    return stack[0]

num = ["2","3","4","*","+"]
print(ope(num))

```

```python
def incentive_efficient(N):
    if not (1 <= N <= 90):
        raise ValueError("Number of days worked must be between 1 and 90.")

    total_incentive = 0

    if N > 50:
        total_incentive = (N - 50) * 1000 + 10 * 800 + 10 * 500
    elif N > 40:
        total_incentive = (N - 40) * 800 + 10 * 500
    elif N > 30:
        total_incentive = (N - 30) * 500

    return total_incentive

days_worked_1 = 28
print(f"For {days_worked_1} days worked, the efficient incentive is: {incentive_efficient(days_worked_1)}")

days_worked_2 = 55
print(f"For {days_worked_2} days worked, the efficient incentive is: {incentive_efficient(days_worked_2)}")

days_worked_3 = 42
print(f"For {days_worked_3} days worked, the efficient incentive is: {incentive_efficient(days_worked_3)}")
```

```python
def find_min_diff_index(n, arr):
  """ Finds the 1-based index of the element with the minimum difference from the array's mean."""

  total_sum = sum(arr)
  mean = total_sum / n

  min_difference = float('inf')
  result_index = -1

  for i in range(n):
    current_difference = abs(arr[i] - mean)

    if current_difference < min_difference:
      min_difference = current_difference
      result_index = i

  return result_index

n_sample = 5
arr_sample = [5,6,7,23,23]
result = find_min_diff_index(n_sample, arr_sample)
print(f"For the array {arr_sample}, the result is: {result}")
```

```python
def can_form_palindrome(nodes):
  """ Checks if the given list of node values can be rearranged to form a palindrome.  """

  counts = [0] * 10

  for node_value in nodes:
    counts[node_value] += 1

  odd_counts = 0
  for count in counts:

    if count % 2 != 0:
      odd_counts += 1

  if odd_counts <= 1:
    return "YES"
  else:
    return "NO"



nodes_sample = [1, 2, 2, 1,7,8]
print(can_form_palindrome(nodes_sample))
```

```python
# Greatest first
from collections import Counter

def sort_greatest_first(arr):
  """Sorts a list of integers based on frequency (descending) and then by first appearance index (ascending)."""
  counts = Counter(arr)

  first_indices = {}
  for i, num in enumerate(arr):
    if num not in first_indices:
      first_indices[num] = i

  sorted_arr = sorted(arr, key=lambda x: (-counts[x], first_indices[x]))

  return sorted_arr

# Example usage with problem's input format
input_str = "1 1 1 1 2 2 2 3 3 4 4 5 5 5 5 5"
numbers = list(map(int, input_str.split()))
result = sort_greatest_first(numbers)
print(' '.join(map(str, result)))
```

```python
# Mid of Linked list
class Node:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def find_middle_node(head: Node) -> int:
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow.val

n = int(input())
values = [int(input()) for _ in range(n)]

# Create the linked list
head = Node(values[0])
current = head
for val in values[1:]:
    current.next = Node(val)
    current = current.next

# Print the middle node value
print(find_middle_node(head))
```
